# Problema: Realizar un sistema que identifique si una planta necesita riego para activar el sensor. 

```python
print (" Sistema de riego autonomo")
Tipo_de_planta = input ("Que planta necesita regar")
Nivel_de_humedad = float(input ("Cual es nivel de humedad en su planata"))
```

```python
print ("Datos almacenados")
print ("Tipo de Planta: ", Tipo_de_planta)
print ("Nivel de humedad: ", Nivel_de_humedad)
```

```python
if Nivel_de_humedad <40:
    print("Necesita Riego")
elif Nivel_de_humedad >=40:
    print("No Necesita Riego")
```
