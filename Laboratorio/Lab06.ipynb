```python
# Importar las librerias

#Libreria utilizada para graficar
import matplotlib.pyplot as plt 
#Libreria que me permite trabajaar con estructuras complejas, como Grafos
import networkx as nx 
```

```python
#Crear una instancia de la clase Graph
#Graph es una colexion de nodos 
grafo = nx.Graph()
```

```python
# Creamos el grafo utiliznado el método add_edges
grafo.add_edges_from(
    [
        ("A","B"),
        ("A","C"),
        ("B","D"),
        ("D","F"),
        ("F","E"),
        ("E","G"),
        ("G","C"),
        ("G","J"),
        ("F","H"),
        ("H","I"),
        ("I","J")
    ]
)
```

```python
# Mostrar el grafo
nx.draw(
    grafo,
    with_labels = True,
    node_size=1000,
    node_color="#e44b8d",
    node_shape="d"
)
plt.show()
```

```python
#Mostrar las rutas posibles del grafo
#Source = De donde inicia
#Target = El objetivo
rutas = list(
    nx.all_simple_paths(grafo, source="A", target="J")
)
```

```python
#Imprimir las rutas
print("Rutas encontradas: ")
for i, ruta in enumerate(rutas, start=1):
    print(f"Ruta {i}: {ruta}")
```

```python
#Mostrar el total de pasos para cada ruta 
pasos1 = len(rutas[0])-1
print("Pasos de la ruta 1: ", pasos1)
pasos2 = len(rutas[1])-1
print("Pasos de la ruta 2: ", pasos2)
pasos3 = len(rutas[2])-1
print("Pasos de la ruta 3: ", pasos3)
pasos4 = len(rutas[3])-1
print("Pasos de la ruta 4: ", pasos4)
```

```python
#Aplicamos la logica proposicional
A = pasos1 < pasos2 and pasos1 < pasos3 and pasos1 < pasos4
B = pasos2 < pasos1 and pasos2 < pasos3 and pasos2 < pasos4 
C = pasos3 < pasos1 and pasos3 < pasos2 and pasos3 < pasos4 
D = pasos4 < pasos1 and pasos4 < pasos2 and pasos4 < pasos3

if A==True:
    print("A = True")
    print("La ruta 1 es la mejor opcion")
elif B==True:
    print("B = True")
    print("La ruta 2 es la mejor opcion")
elif C==True:
    print("C = True")
    print("La ruta 3 es la mejor opcion")
elif D==True:
    print("D = True")
    print("La ruta 4 es la mejor opcion")
else:
    print("A = False")
    print("B = False")
    print("C = False")
    print("D = False")
    print("Las dos rutas son igual de buenas")
    

    
```
